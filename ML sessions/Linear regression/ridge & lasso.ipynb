# Cross validation
### this method is for choosing the best lambda -l- , -alpha-

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score,KFold
import numpy as np

#alpha is a lambda
alphas=[.001,.01,.1,.2,.3,1,2,3,10,100]

#this is what folds data into sets
                    #for random shuffling
kf=KFold(n_splits=3,shuffle=True           ,random_state=1)

for alpha in alphas:
    rmodel=Ridge(alpha=alpha)
    
    #the line that calculate the model performance by the cross validation
    #its better when the number is near to one (1)
                                        #the folding method  the method for scoring      
    scores = cross_val_score(rmodel,x,y,cv=kf               ,scoring='r2')
    print(f'lambda = {alpha} performance is {np.mean(scores):.3f}')
```

## The ElasticNet
combination between the Ridge and Lasso

```python
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import cross_val_score, KFold
import numpy as np

# نختار قيم مختلفة للـ alpha
alphas = [0.001, 0.01, 0.1, 1, 10]

# نعمل الـ KFold
kf = KFold(n_splits=5, shuffle=True, random_state=1)

for alpha in alphas:
                                   #the percentage between the ridge and lasso
                                   #0 Ridge  1 Lasso
    model = ElasticNet(alpha=alpha, l1_ratio=0.5,                            random_state=1)
    scores = cross_val_score(model, x, y, cv=kf, scoring='r2')
    print(f'alpha={alpha} --> performance={np.mean(scores):.3f}')
```
